# Eşik Ayarlaması

👇 Çalışacağınız veri setini görmek için `player_performances.csv` oyuncu veri setini yükleyin.

```python

import pandas as pd
from pathlib import Path

url = "https://d32aokrjazspmn.cloudfront.net/materials/ML_Player_performance.csv"

data_path = Path("data/player_performances.csv")
data_path.parent.mkdir(parents=True, exist_ok=True)

if not data_path.exists():
    pd.read_csv(url).to_csv(data_path, index=False)

data = pd.read_csv(data_path)

print("Veri setinin boyutu:", data.shape)
display(data.head())

print("\nHedef sınıf dağılımı:")
print(data["target_5y"].value_counts())
```

ℹ️ Her gözlem bir oyuncuyu temsil eder ve her sütun performansın bir özelliğidir. Hedef `target_5y`, oyuncunun 5 yıldan az [0] veya 5 yıl ve daha fazla [1] profesyonel kariyere sahip olup olmadığını tanımlar.

# Ön İşleme

👇 Ön işleme için çok fazla zaman harcamamak adına, tüm özellik setini Robust Scale ile ölçeklendirin. Bu uygulama optimal değildir, ancak ön işleme ve/veya modellerin hızla çalıştırılması için kullanılabilir.

Ölçeklendirilmiş özellik setini `X_scaled` olarak kaydedin.

```python

from sklearn.preprocessing import RobustScaler
import pandas as pd

# Hedef değişkeni özelliklerden ayır
X = data.drop(columns=["target_5y"])
y = data["target_5y"]

# RobustScaler oluştur
scaler = RobustScaler()

# Özellikleri ölçeklendir
X_scaled = pd.DataFrame(
    scaler.fit_transform(X),
    columns=X.columns,
    index=X.index
)

print("Ölçeklendirilmiş veri boyutu:", X_scaled.shape)

display(X_scaled.head())
```

### ☑️ Kodunuzu kontrol edin

```python
from nbresult import ChallengeResult

result = ChallengeResult('scaled_features',
                         scaled_features = X_scaled
)

result.write()
print(result.check())
```

# Temel Modelleme

🎯 Görev, %90 garantiyle minimum 5 yıl profesyonel olarak devam edecek oyuncuları tespit etmektir.

👇 Varsayılan bir Lojistik Regresyon modeli antrenörün gereksinimlerini karşılayacak mı? Çapraz doğrulama kullanın ve cevabınızı destekleyen skoru `base_score` değişken adı altında kaydedin.

```python

from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import cross_val_score, StratifiedKFold

# Logistic Regression modelini oluştur
model = LogisticRegression(max_iter=2000)

# 5 katlı çapraz doğrulama
cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

# Precision skorlarını hesapla
precision_scores = cross_val_score(
    model,
    X_scaled,
    y,
    cv=cv,
    scoring="precision"
)

# Ortalama Precision
base_score = precision_scores.mean()

print("Fold Precision Scores:", precision_scores)
print(f"Baseline Precision: {base_score:.4f}")
print(f"Baseline Precision (%): {base_score * 100:.2f}")

if base_score >= 0.90:
    print("Model %90 Precision hedefine ulaşıyor.")
else:
    print("Model %90 Precision hedefine ulaşamıyor.")
```

### ☑️ Kodunuzu kontrol edin

```python
from nbresult import ChallengeResult

result = ChallengeResult('base_precision',
                         score = base_score
)

result.write()
print(result.check())
```

# Eşik Ayarlaması

👇 Bir oyuncunun profesyonel olarak 5 yıl veya daha fazla sürmesi için %90 kesinlik garantisi veren karar eşiğini bulun. Eşiği `new_threshold` değişken adı altında kaydedin.

<details>
<summary>💡 İpucu</summary>

- [`cross_val_predict`](https://scikit-learn.org/stable/modules/generated/sklearn.model_selection.cross_val_predict.html) ile çapraz doğrulanmış olasılık tahminleri yapın
    
- Farklı eşiklerde kesinlik skorları oluşturmak için olasılıkları [`precision_recall_curve`](https://scikit-learn.org/stable/modules/generated/sklearn.metrics.precision_recall_curve.html) içine yerleştirin

- 0.9 kesinliği garanti eden eşiği bulun
      
</details>

```python

import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import cross_val_predict
from sklearn.metrics import precision_recall_curve

# Çapraz doğrulama ile olasılık tahminleri
y_prob = cross_val_predict(
    model,
    X_scaled,
    y,
    cv=cv,
    method="predict_proba"
)[:, 1]

# Precision - Recall eğrisi
precisions, recalls, thresholds = precision_recall_curve(
    y,
    y_prob
)

# %90 Precision sağlayan eşikleri bul
valid_indices = np.where(precisions[:-1] >= 0.90)[0]

if len(valid_indices) == 0:
    raise ValueError("Bu modelde %90 Precision sağlayan eşik bulunamadı.")

# %90 Precision sağlayan en düşük eşik
new_threshold = float(thresholds[valid_indices[0]])

print(f"Yeni karar eşiği: {new_threshold:.4f}")
print(f"Precision: {precisions[valid_indices[0]]:.4f}")
print(f"Recall: {recalls[valid_indices[0]]:.4f}")

# Grafiği çiz
plt.figure(figsize=(10, 6))

plt.plot(thresholds, precisions[:-1], label="Precision")
plt.plot(thresholds, recalls[:-1], label="Recall")

plt.axhline(
    y=0.90,
    color="red",
    linestyle="--",
    label="%90 Precision hedefi"
)

plt.axvline(
    x=new_threshold,
    color="green",
    linestyle="--",
    label=f"Yeni eşik: {new_threshold:.3f}"
)

plt.xlabel("Karar Eşiği (Threshold)")
plt.ylabel("Skor")
plt.title("Precision - Recall vs Threshold")
plt.legend()
plt.grid(True)
plt.show()
```

### ☑️ Kodunuzu kontrol edin

```python
from nbresult import ChallengeResult

result = ChallengeResult('decision_threshold',
                         threshold = new_threshold
)

result.write()
print(result.check())
```

# Yeni Eşiği Kullanma

🎯 Antrenör potansiyel olarak ilginç bir oyuncu fark etti, ancak bu oyuncunun minimum 5 yıl profesyonel olarak devam edeceğine dair %90 garantinizi istiyor. Oyuncunun verilerini [buradan](https://wagon-public-datasets.s3.amazonaws.com/Machine%20Learning%20Datasets/ML_New_player.csv) indirin.

```python

new_player = pd.read_csv(
    "https://d32aokrjazspmn.cloudfront.net/materials/ML_New_player.csv"
)

new_player
```

❓ Oyuncuyu antrenöre tavsiye etmeyi göze alır mısınız? Cevabınızı string olarak `recommendation` değişken adı altında "recommend" veya "not recommend" şeklinde kaydedin.

```python

# Modeli tüm eğitim verisiyle eğit
model.fit(X_scaled, y)

# Yeni oyuncunun özelliklerini sırala
new_player_X = new_player.reindex(columns=X.columns)

# Önceki RobustScaler ile ölçeklendir
new_player_scaled = scaler.transform(new_player_X)

# Oyuncunun 5+ yıl oynama olasılığı
player_probability = model.predict_proba(
    new_player_scaled
)[0, 1]

# Yeni karar eşiğini uygula
recommendation = (
    "recommend"
    if player_probability >= new_threshold
    else "not recommend"
)

print(f"Oyuncunun tahmini olasılığı: %{player_probability * 100:.2f}")
print(f"Karar eşiği: %{new_threshold * 100:.2f}")
print("Antrenöre tavsiye:", recommendation)
```

### ☑️ Kodunuzu kontrol edin

```python
from nbresult import ChallengeResult

result = ChallengeResult('recommendation',
                         recommendation = recommendation
)

result.write()
print(result.check())
```
